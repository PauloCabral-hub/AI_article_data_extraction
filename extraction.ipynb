```python
# Packages
import os
import subprocess
from pathlib import Path
from docx import Document
from mistralai.client import Mistral
import pandas as pd
```

```python
# Library
def pdf_to_text(pdf_path: str, txt_path: str) -> None:
    """Convert a PDF file to text using pdftotext."""
    subprocess.run(
        ["pdftotext", "-layout", pdf_path, txt_path],
        check=True,              # raises CalledProcessError on failure
        capture_output=True,
        text=True,
    )

# usage example:
# pdf_to_text("input.pdf", "output.txt")
```

```python
# Convertion from .pdf to .txt
root_folder = '/home/paulocabral/Documents/AI_article_data_extraction/'

target_orig_folder = "pdfs_to_extract"
target_dest_folder = "pdfs_extracted"

target_pdf = "Capogrosso2013.pdf"
target_txt = target_pdf.split(".")[0]

complete_orig = os.path.join(root_folder, target_orig_folder)
complete_orig = os.path.join(complete_orig, target_pdf)
#
complete_dest = os.path.join(root_folder, target_dest_folder)
complete_dest = os.path.join(complete_dest, target_txt)

pdf_to_text(complete_orig, complete_dest)
```
